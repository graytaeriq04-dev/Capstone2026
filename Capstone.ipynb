```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import dataquality as dq
```

```python
df = pd.read_csv(r'cc_data.csv')
df
```

```python
dq.data_check(df)
```

```python
dfcorr = df.corr(numeric_only=True)
dfcorr
```

1. is_fraud had the most correlation to amt, however it's still realativly weak. While I'd assume fruadsters to make larger transactions not all large transactions are fraud.
2. The latitude and longitude of has pretty large correlation to the merchants latitude and longitude meaning when fruad happens it's more than likely local, however some of the fraud happens online but it would seem more fraud happens in person rather than online.

```python
plt.figure(figsize=(12,10))
sns.heatmap(dfcorr, annot=True, cmap='coolwarm')
plt.savefig("Correlation.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
sns.kdeplot(data=df, x='amt', hue='is_fraud', common_norm=False)
plt.xlim(0, 2000)
plt.savefig("kdeplot.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
fig, axes = plt.subplots(2, 1, figsize=(12, 8))

legit = df[df['is_fraud'] == 0]
fraud = df[df['is_fraud'] == 1]

for ax, col, ylabel in zip(axes, ['amt', 'city_pop'], ['Amount ($)', 'City Population']):
    ax.scatter(legit.index, legit[col], alpha=0.2, s=5, color='steelblue', label='Legitimate')
    ax.scatter(fraud.index, fraud[col], alpha=0.6, s=10, color='red', label='Fraud')
    ax.set_xlabel('Transaction Index')
    ax.set_ylabel(ylabel)

axes[0].set_title('Transaction Amount — Fraud vs Legitimate')
axes[1].set_title('City Population — Fraud vs Legitimate')
fig.legend(*axes[0].get_legend_handles_labels(), loc='upper right')

plt.tight_layout()
plt.savefig("scatter_fraud_v2.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
df['trans_date_trans_time'] = pd.to_datetime(df['trans_date_trans_time'])
df['hour'] = df['trans_date_trans_time'].dt.hour
df['trans_date_trans_time'] = pd.to_datetime(df['trans_date_trans_time'], dayfirst=True)
fraud_by_hour = df[df['is_fraud'] == 1].groupby('hour')['is_fraud'].count()
fraud_by_hour
```

```python
fig, ax = plt.subplots(figsize=(12, 6))

ax.bar(fraud_by_hour.index, fraud_by_hour.values)
ax.set_xlabel('Hour of Day (24hr)')
ax.set_ylabel('Number of Fraudulent Transactions')
ax.set_title('Fraud by Hour of Day')
ax.set_xticks(range(0, 24))

plt.savefig("fraud_by_the_hour_v2.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
transactions = {
    "Hudson-Grady\n(shopping_pos)": {"amt": 2.42, "lat": 45.176, "lon": -92.835},
    "Kemmer-Buckridge\n(misc_pos)":  {"amt": 6.16, "lat": 45.324, "lon": -94.129},
}

fig, ax = plt.subplots(figsize=(10, 5))
# Plot merchants as points on a lat/lon scatter
for name, t in transactions.items():
    ax.scatter(t["lon"], t["lat"], s=200, zorder=5)
    ax.annotate(f"{name}\n${t['amt']}", (t["lon"], t["lat"]),
                textcoords="offset points", xytext=(10, 5), fontsize=9)

ax.set_title("Same card · Same timestamp · Two locations\n4933 4619 3034 8830 — 2019-04-14 16:33:00",
             color="red", fontweight="bold")
ax.set_xlabel("Longitude")
ax.set_ylabel("Latitude")
plt.tight_layout()
plt.savefig("fraud_map.png", dpi=300)
plt.show()
```
